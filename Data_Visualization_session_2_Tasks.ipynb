## Task 5: Detect Possible Outliers

Use these numerical columns:

- `alcohol`
- `malic_acid`
- `color_intensity`
- `proline`

**Requirements:**

1. Create a Box Plot for each feature.
2. Compare the distributions.
3. Calculate Q1 and Q3 for each feature.
4. Calculate the Interquartile Range (IQR).
5. Identify possible outliers using the 1.5 × IQR rule.
6. Count the possible outliers in each feature.
7. Decide whether the outliers should be kept, removed, or investigated.

**Questions:**
- Which features contain possible outliers?
- Does an outlier always mean an incorrect value?
- Why can removing outliers be risky?

**Important:** Do not automatically remove all outliers. Some extreme values may be valid chemical measurements.


```python
import matplotlib.pyplot as plt

features = df.columns

for col in features:
    df.boxplot(column=col)
    plt.title(f"Box Plot of {col}")
    plt.show()
```

```python
results = []

for col in features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)][col]

    results.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower,
        "Upper Bound": upper,
        "Possible Outliers": len(outliers)
    })

outlier_table = pd.DataFrame(results)
outlier_table.round(2)
```

### 1. Features containing possible outliers can be identified from the table.
### 2. No, an outlier does not always mean an incorrect value.
### 3. Removing outliers can discard valid information and introduce bias Decision: Investigate possible outliers before removing them.
